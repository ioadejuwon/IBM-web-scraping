```python
import pandas as pd
import csv, sqlite3 
import prettytable
prettytable.DEFAULT = 'DEFAULT'

conn = sqlite3.connect("FinalDB.db")
cur = conn.cursor()
```

```python
%load_ext sql
%sql sqlite:///FinalDB.db
```

```python
df = pd.read_csv("../data/ChicagoCensusData.csv")
df.to_sql("CENSUS_DATA", conn, if_exists='replace', index=False, method="multi")

df2 = pd.read_csv("../data/ChicagoCrimeData.csv")
df2.to_sql("CHICAGO_CRIME_DATA", conn, if_exists='replace', index=False, method="multi")

df3 = pd.read_csv("../data/ChicagoPublicSchools.csv")
df3.to_sql("CHICAGO_PUBLIC_SCHOOLS", conn, if_exists='replace', index=False, method="multi")
```
